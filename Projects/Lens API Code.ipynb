```python
import requests
import json
# Make an API call and store the response.
url = 'https://api.lens.org/patent/search'
query = {
  "query": {
    "match": {
      "title": "Aeroplane AND drag*"
             }
           },
  "size": 5,
  "from": 0,
  "include": [
    "abstract"
  ],
  "sort": [
    {
      "created": "desc"
    },
    {
      "year_published": "asc"
    }
  ],
  "exclude": None,
  "scroll": None,
  "scroll_id": None,
  "Stemming": False,
  "group_by": "SIMPLE_FAMILY"
}
headers = {'Authorization': 'Bearer l3RL63QN7E2X0zV0pxKrguKN5yxvBYtiG3G6MGlXtksmf4BUkR4K', 'Content-Type': 'application/json'}
r = requests.post(url, headers=headers, json=query)
print("Status code:", r.status_code)
with open("lens_results.json", "w") as f:
        json.dump(r.json(), f, indent=4)
print("Saved lens_results.json")
```
